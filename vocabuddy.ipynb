```python
# ----------------------------------------
# 組員一：新增單字庫內容
# ----------------------------------------

vocab = {
    "repository": "倉庫",
    "invitation": "邀請",
    "collaborate": "合作",
    "commit": "提交",
    "branch": "分支",
    "merge": "合併",
}

print("=== 目前單字庫內容 ===")
for eng, chi in vocab.items():
    print(f"{eng} : {chi}")

# ----------------------------------------
# 組員二：新增單字統計與隨機測驗功能
# ----------------------------------------
import random

# 1. 顯示單字總數
print(f"目前單字庫共有 {len(vocab)} 個單字。\n")

# 2. 隨機單字測驗功能
def vocabulary_quiz(vocab_dict, num_questions=3):
    correct_count = 0
    wrong_count = 0

    # 隨機抽取指定題數（不重複）
    quiz_items = random.sample(list(vocab_dict.items()), min(num_questions, len(vocab_dict)))

    print("====== 單字測驗開始 ======")
    for index, (eng, chi) in enumerate(quiz_items, 1):
        answer = input(f"第 {index} 題 - 請輸入「{chi}」的英文：").strip().lower()

        if answer == eng.lower():
            print(" 答對了！\n")
            correct_count += 1
        else:
            print(f" 答錯了！正確答案是：{eng}\n")
            wrong_count += 1

    print("====== 測驗結果 ======")
    print(f"總共測驗：{len(quiz_items)} 題")
    print(f"答對題數：{correct_count} 題")
    print(f"答錯題數：{wrong_count} 題")
    accuracy = (correct_count / len(quiz_items)) * 100
    print(f"正確率：{accuracy:.1f}%")

# 執行 3 題隨機測驗
vocabulary_quiz(vocab, num_questions=3)
```
